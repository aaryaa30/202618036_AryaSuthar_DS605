TASK 2 - Vectorized Arithmetic and Linear Algebra

```python
# Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().
A = np.array([
    [2, 4],
    [6, 8]
])

B = np.array([
    [1, 3],
    [5, 7]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("Addition:")
print(A + B)

print("\nSubtraction:")
print(A - B)

print("\nElement-wise multiplication:")
print(A * B)
```

```python
#matrix multiplication
print("Matrix multiplication:")
print(A @ B)
```

```python
#transpose and determininant
print("Transpose of A:")
print(A.T)

det_A = np.linalg.det(A)

print("\nDeterminant of A:")
print(det_A)
```

```python
#If invertible, find the inverse and verify A @ A^-1 using np.allclose().
if not np.isclose(det_A, 0):
    A_inverse = np.linalg.inv(A)
    print("Inverse of A:")
    print(A_inverse)
else:
    print("A is not invertible.")
```

```python
#verifyingidentity_result = A @ A_inverse
identity_result = A @ A_inverse

print("A × A inverse:")
print(identity_result)

print("\nIs the result approximately an identity matrix?")
print(np.allclose(identity_result, np.eye(A.shape[0])))
```

TASK 5- Filtering and Querying

```python
#Male passengers older than 50: how many?
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("Male passengers older than 50:")
print(male_over_50)

print("\nNumber of male passengers older than 50:", len(male_over_50))
```

```python
#Female first-class passengers: how many, and what percentage survived?
female_first_class = df[
    (df["Sex"] == "female") & 
    (df["Pclass"] == 1)
]

print("Number of female first-class passengers:", len(female_first_class))

survival_percentage = female_first_class["Survived"].mean() * 100

print("Percentage survived:", survival_percentage)
```

```python
#Age 20-40, Fare above overall median, and survived: how many?
median_fare = df["Fare"].median()

print("Overall median fare:", median_fare)

condition_3 = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > median_fare) &
    (df["Survived"] == 1)
]

print("\nPassengers aged 20-40, with fare above overall median, who survived:")
print(condition_3)

print("\nNumber of passengers:", len(condition_3))
```

```python
#Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
condition_4 = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("Passengers travelling alone, age < 30, who did not survive:")
print(condition_4)

print("\nNumber of passengers:", len(condition_4))
```

```python
#Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?
southampton_median_fare = df.loc[
    df["Embarked"] == "S",
    "Fare"
].median()

print("Southampton median fare:", southampton_median_fare)
```

```python
condition_5 = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

print("Passengers satisfying the condition:")
print(condition_5)

print("\nNumber of passengers:", len(condition_5))
```

TASK-6 groupby() and Aggregation

```python
#Survival rate by Sex.
survival_by_sex = df.groupby("Sex")["Survived"].mean()

print("Survival rate by Sex:")
print(survival_by_sex)
```

```python
#Survival rate by Pclass.
survival_by_pclass = df.groupby("Pclass")["Survived"].mean()

print("Survival rate by Pclass:")
print(survival_by_pclass)
```

```python
#Average Age and Fare by Pclass.
avg_age_fare = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("Average Age and Fare by Pclass:")
print(avg_age_fare)
```

```python
#Passenger count and survival rate by Sex-Pclass.
sex_pclass = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

print("Passenger count and survival rate by Sex-Pclass:")
print(sex_pclass)
```

```python
#Passenger count, average Fare, and survival rate by Embarked.
embarked_summary = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

print("Passenger count, average Fare, and survival rate by Embarked:")
print(embarked_summary)
```

TASK -7 Missing Values and Fare Outliers

```python
#Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.
missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

print(missing_summary)
```

```python
#plotting 
plt.figure(figsize=(10, 5))

missing_count.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Count")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("missing_values.png")
plt.show()
```

```python
#Fill missing Age values with the mean Age. Show missing counts before and after.
print("Missing Age before:", df["Age"].isnull().sum())

mean_age = df["Age"].mean()

df["Age"] = df["Age"].fillna(mean_age)

print("Missing Age after:", df["Age"].isnull().sum())
```

```python
#filling NaN values by imputing using: mean, median, mode, random value
#mean
age_mean = df["Age"].fillna(df["Age"].mean())

#median
age_median = df["Age"].fillna(df["Age"].median())

#mode
age_mode = df["Age"].fillna(df["Age"].mode()[0])

#ranfom value
age_random = df["Age"].copy()

missing_indices = age_random[age_random.isnull()].index

random_values = df["Age"].dropna().sample(
    n=len(missing_indices),
    random_state=42
).values

age_random.loc[missing_indices] = random_values

print("Mean imputation completed.")
print("Median imputation completed.")
print("Mode imputation completed.")
print("Random-value imputation completed.")
```

```python
# calculating Q1, Q3, IQR, 1.5xIQR bounds, and counting the outliers.
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
#counting the outlier
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Number of Fare outliers:", len(fare_outliers))
```

TASK-8 Features and Pivot Table

```python
#Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10))
```

```python
#Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest groups.
pivot_table = pd.pivot_table(
    df,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print("Mean Survival Rate by Sex and Passenger Class:")
print(pivot_table)
```

```python
#finding the highest and lowest survival groups
highest_group = pivot_table.stack().idxmax()
highest_rate = pivot_table.stack().max()

lowest_group = pivot_table.stack().idxmin()
lowest_rate = pivot_table.stack().min()

print("Highest survival group:", highest_group)
print("Highest survival rate:", highest_rate)

print("\nLowest survival group:", lowest_group)
print("Lowest survival rate:", lowest_rate)
```
